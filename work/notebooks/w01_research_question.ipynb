## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
Research Question: Can we predict which content pages are underperforming relative to their search impression potential and identify pages likely to see at least a 15% increase in CTR if rewritten?

Unit of Analysis: A single content URL aggregated over a 30-day window.

Target Output: Binary classification (1 = High Potential for CTR Growth, 0 = Low Potential / Stable).

Actionable Decision: Content managers use model predictions to select specific high-potential pages for weekly copy updates, title optimization, and internal linking strategies.

Cost of Misclassification/Error:

False Positive (Cost of Waste): Spending $300–$500 per article rewriting pages that fail to gain additional traffic, wasting limited editorial budget and time.
False Negative (Cost of Opportunity): Leaving high-potential decaying content untouched, causing the site to permanently lose search rankings and high-intent organic visitors to competitors.
```
