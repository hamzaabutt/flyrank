## 1. Question

**Research Question:**
Can we predict organic traffic decay (`is_declining`) across large content catalogs using pre-decision historical signals, and construct a ranked action playbook that reliably prioritizes high-opportunity URLs for editorial refresh?

**Decision Supported:**
Content strategists allocate $300–$500 refresh budgets per article. This model provides empirical decision-support scoring to prioritize high-yield pages and avoid wasting editorial resources on stable or low-demand content.

```python
import os, sys, json
from pathlib import Path
import pandas as pd
import numpy as np

# Set path relative to repo root
repo_root = Path("../../").resolve() if Path("../../scripts").exists() else Path("../").resolve()
sys.path.append(str(repo_root / "scripts"))
from ml_utils import PROCESSED_DIR, OUTPUT_DIR

# Load model results receipt
results_path = OUTPUT_DIR / "model_results.json"
if not results_path.exists():
    results_path = repo_root / "outputs" / "model_results.json"

with open(results_path, "r") as f:
    results = json.load(f)

print(f"Capstone Analysis Initialized:")
print(f"- Sample Size: {results['input_rows']:,} anonymized content URLs")
print(f"- Target: {results['target']} (Positives: {results['target_positive_rate']:.1%})")
```

## 2. Data

**Data Specs & Exclusions (Public-Safe):**
- **Source**: FlyRank anonymized search intelligence dataset (`content_refresh_anonymized.csv`).
- **Scale**: 30,000 unique content URLs across multiple client domains.
- **Exclusions**: Pages with zero 90-day impressions (`impressions_90d == 0`) or content age $< 90$ days were excluded to focus on mature, indexable assets.
- **Privacy**: All domain names, client IDs, and query strings are fully anonymized.

```python
print(f"Total Cleaned Dataset Rows: {results['input_rows']:,}")
print(f"Train Rows (80% Client Holdout): {results['train_rows']:,}")
print(f"Test Rows (20% Client Holdout): {results['test_rows']:,}")
```

## 3. Methodology

**Feature Matrix & Validation Design:**
- **Features**: 52 numerical and categorical features (`days_with_impressions`, `log_impressions_90d`, `avg_position`, `content_age_days`, `word_count`, `ctr`, etc.).
- **Target Label**: `is_declining_label` ($1 = \text{downward trend}$, $0 = \text{stable/upward}$). Defined strictly from historical trend classification.
- **Validation Split**: `client_holdout` split — 80% client domains for training, 20% independent client domains held out for testing to prevent domain leakage.
- **Leakage Prevention**: All features are computed strictly prior to label observation.

```python
fi_df = pd.DataFrame(results["best_model"]["feature_importance_top"]).head(8)
print("Top Feature Importance Weights (Random Forest):")
fi_df
```

## 4. Results (vs baseline)

**Empirical Results Table (Evaluated on 2,325 Held-out Test URLs):**

| Model | Precision@50 | Precision@20 | PR-AUC | ROC-AUC | Accuracy |
|---|---|---|---|---|---|
| **Deterministic Baseline (Week 4)** | 24.0% | 15.0% | 0.468 | 0.627 | 60.9% |
| **Logistic Regression** | 40.0% | 35.0% | 0.522 | 0.700 | 66.1% |
| **Decision Tree** | 62.0% | 55.0% | 0.575 | 0.742 | 67.7% |
| **Random Forest (Best Model)** | **74.0%** | **65.0%** | **0.618** | **0.750** | **67.2%** |

*The Random Forest achieves a **3.08x lift** in Precision@50 over the baseline rule.*

```python
best_metrics = results["models"][results["best_model"]["name"]]
base_metrics = results["baseline"]
lift = best_metrics['precision_at_50'] / base_metrics['baseline_precision_at_50']
print(f"Random Forest P@50: {best_metrics['precision_at_50']:.1%}")
print(f"Baseline P@50: {base_metrics['baseline_precision_at_50']:.1%}")
print(f"Precision@50 Lift: {lift:.2f}x")
```

## 5. Limitations

**Honest Framing & Limits:**
- **Decision Support, Not Causality**: Predictions indicate statistical associations with historical traffic decay; they do not guarantee algorithmic ranking changes by search engines.
- **Observational Data**: Results rely on historical impression and click distributions.
- **Domain Scope**: Evaluated across anonymized content sites; niche industries may require domain-specific calibration.

## 6. Ranked recommendations

**Action Playbook Engine:**
- **`expand_and_refresh`**: Thin visible pages (`word_count < 1200`). Add key subtopics and structured FAQs.
- **`refresh_and_review_ctr`**: Low CTR visible pages (`ctr < 0.5%`). Rewrite title tags and optimize meta descriptions.
- **`refresh`**: Decaying or stale pages (`days_since_last_update >= 180`). Update statistics and publish dates.
